# 01 - COVID-19 Data Overview

This notebook provides an initial overview of the COVID-19 dataset from Our World in Data (OWID).

## Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

# Import the data loader module
from src.data_loader import load_csv, get_basic_info
from src.visualization import set_style

# Set visualization style
set_style()

print('Libraries imported successfully!')
```

## Load Data

```python
# Load the COVID-19 dataset
data_path = 'data/raw/compact.csv'
df = load_csv(data_path)
```

## Data Overview

```python
# Display basic information
print('Dataset Shape:', df.shape)
print('\nFirst few rows:')
df.head()
```

```python
# Get detailed information
info = get_basic_info(df)
print('Column Names:', info['columns'])
print('\nData Types:')
for col, dtype in info['dtypes'].items():
    print(f'  {col}: {dtype}')
```

```python
# Check for missing values
print('Missing Values Summary:')
missing_info = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing_info,
    'Percentage': missing_pct
})
print(missing_df[missing_df['Missing Count'] > 0].sort_values('Percentage', ascending=False))
```

## Summary Statistics

```python
# Display summary statistics
print('Summary Statistics:')
print(df.describe())
```

```python
# Check unique values in key columns
print('Unique Countries:', df['location'].nunique() if 'location' in df.columns else 'N/A')
print('Date Range:', df['date'].min(), 'to', df['date'].max() if 'date' in df.columns else 'N/A')
```
